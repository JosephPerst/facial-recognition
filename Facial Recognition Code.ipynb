```python
maximumface = 0

for key in data.keys():
    if len(data[key]) > maximumface:
        maximumface = len(data[key])
        maximumfacepic = key
print("The picture with the most faces is", maximumfacepic, "with", maximumface, "faces")
```

```python
import numpy as np

zendaya = []
with open('face_data/zendaya.txt', 'r') as infile:
    for line in infile:
        zendaya.append(float(line.rstrip('\n')))
    
zendaya = np.array(zendaya)
```

```python
def euclidean_distance(vector1, vector2):
    return np.sqrt(np.sum((vector1 - vector2) ** 2))


# Initialize a counter for matches
zendayaamount = 0

# Iterate through each picture in the dataset to 
for key, faces in data.items():
    for face_vector in faces:
        # Convert theface vectors in each iteration to a numpy array
        vector = np.array(face_vector)
        # Calculate the Euclidean distance
        distance = euclidean_distance(vector, zendaya)
        # If the distance is less than 0.5 it is added to the count
        if distance < 0.5:
            zendayaamount += 1

print("Zendaya appears", zendayaamount, "times.")
```
